## 1. Loading the dataset

```python
import os
from monai.apps import DecathlonDataset
from monai.transforms import Compose, LoadImaged, EnsureChannelFirstd

# Carpeta on es descarregaran les dades (ha d'existir abans)
os.makedirs("data", exist_ok=True)

# Cadena de transformacions que s'aplica a cada mostra quan la demanem
transform = Compose([
    LoadImaged(keys=["image", "label"]),          # llegeix els fitxers NIfTI
    EnsureChannelFirstd(keys=["image", "label"]),  # (H, W, D) -> (1, H, W, D)
])

ds = DecathlonDataset(
    root_dir="data",
    task="Task04_Hippocampus",
    section="training",
    transform=transform,
    download=True,
    cache_rate=0.0,
    val_frac=0.2,
)

print("Nombre de volums:", len(ds))
```

```python
sample = ds[0]            # aquí es llegeix el fitxer i s'apliquen les transformacions
img = sample["image"][0]  # treiem la dimensió de canal: (1, H, W, D) -> (H, W, D)
lab = sample["label"][0]

print("Mida del volum:", img.shape)
print("Espaiat dels vòxels (mm):", img.meta["pixdim"][1:4])
print("Rang d'intensitats:", float(img.min()), float(img.max()))
print("Etiquetes:", lab.unique())
```

**Observations (first volume):**
- Volume size 34×56×31: these are crops around the hippocampus, not whole brains.
- Isotropic spacing of 1 mm.
- Raw MRI intensities (2–112) have no physical units → per-volume normalization will be needed.
- Labels: 0 = background, 1 = anterior hippocampus, 2 = posterior hippocampus.

## 2. Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

img_np = img.numpy()
lab_np = lab.numpy()

def show_planes(img_np, lab_np):
    """Mostra els tres talls ortogonals pel centre del volum, amb l'etiqueta a sobre."""
    c = [s // 2 for s in img_np.shape]   # índex central de cada eix
    slices = [
        ("Axis 0 fixed", img_np[c[0], :, :], lab_np[c[0], :, :]),
        ("Axis 1 fixed", img_np[:, c[1], :], lab_np[:, c[1], :]),
        ("Axis 2 fixed", img_np[:, :, c[2]], lab_np[:, :, c[2]]),
    ]
    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    for ax, (title, im, lb) in zip(axes, slices):
        ax.imshow(im.T, cmap="gray", origin="lower")
        # Amaguem els vòxels de fons perquè no es pintin
        lb_masked = np.ma.masked_where(lb.T == 0, lb.T)
        ax.imshow(lb_masked, cmap="autumn", alpha=0.5, origin="lower", vmin=1, vmax=2)
        ax.set_title(title)
        ax.axis("off")
    plt.tight_layout()
    plt.show()

show_planes(img_np, lab_np)
```

## 3. Dataset-level statistics

```python
import pandas as pd

rows = []
for i in range(len(ds)):
    s = ds[i]
    im = s["image"][0]
    lb = s["label"][0]
    n = lb.numel()                      # nombre total de vòxels del volum
    sp = im.meta["pixdim"][1:4]

    rows.append({
        "file": os.path.basename(im.meta["filename_or_obj"]),
        "size_0": im.shape[0], "size_1": im.shape[1], "size_2": im.shape[2],
        "spacing_0": float(sp[0]), "spacing_1": float(sp[1]), "spacing_2": float(sp[2]),
        "int_min": float(im.min()), "int_max": float(im.max()),
        "frac_bg": float((lb == 0).sum()) / n,
        "frac_c1": float((lb == 1).sum()) / n,
        "frac_c2": float((lb == 2).sum()) / n,
    })

df = pd.DataFrame(rows)
print(df.describe().round(3))
```

**Observations (dataset level, n = 208):**
- Volume sizes vary (31–43 × 40–59 × 24–47) → a common size is needed for batching.
- Spacing is 1 mm isotropic for every volume.
- Strong class imbalance: background ≈ 94.7%, anterior ≈ 2.7%, posterior ≈ 2.5%.
  Voxel accuracy is meaningless here; Dice will be used instead.
- Maximum intensities range from 112 to ~1.7·10⁶ (median ≈ 2·10³): intensity scales
  differ by orders of magnitude between volumes.

```python
print(df.sort_values("int_max", ascending=False)[["file", "int_min", "int_max"]].head())

# Visualitzem el volum amb el màxim més alt
idx = df["int_max"].idxmax()
s = ds[idx]
show_planes(s["image"][0].numpy(), s["label"][0].numpy())
```
